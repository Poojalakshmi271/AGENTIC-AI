```python
import zipfile

with zipfile.ZipFile("movie-dialog-corpus.zip", 'r') as zip_ref:
    zip_ref.extractall("data/movie-dialog-corpus")
```

```python
import csv
import random

f = "data/movie-dialog-corpus/movie_lines.tsv"

texts = []

with open(f, encoding="utf-8", errors="ignore") as i:

    csv_reader = csv.reader(i, delimiter="\t")

    for line in csv_reader:

        if len(line) > 0:
            texts.append(line[-1])

random.seed(1)
random.shuffle(texts)

texts = texts[:10000]

print(texts[:10])


```

```python
import tensorflow_hub as hub
import tensorflow as tf
import numpy as np




```

```python
tf.compat.v1.logging.set_verbosity(tf.compat.v1.logging.ERROR)

embed=hub.load("https://tfhub.dev/google/universal-sentence-encoder/2")
```

```python
def embed_texts(texts):
    tf.compat.v1.logging.set_verbosity(tf.compat.v1.logging.ERROR)
    with tf.compat.v1.Session() as session:
        session.run([tf.compat.v1.global_variables_initialiser(), tf.compat.v1.tables_initialiser()])
        embeddings=session.run(embed(texts))
    return np.array(embeddings).tolist()
embeddings=embed_texts(texts)
```
