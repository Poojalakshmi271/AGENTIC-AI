Defining the features(X) and Target(Y) for my data

```python
X=df.drop(['Salary','Unnamed: 0'], axis=1)
y=df['Salary']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
model=model=LinearRegression()
model=model.fit(x_train,y_train)
```

```python
predictSalary=model.predict(x_test)
```

```python
predictSalary
```

```python
errors=mean_squared_error(y_test, predictSalary)
print('mean_squared_error',errors)

r2=r2_score(y_test, predictSalary)
print('R2 Score',r2_score)

errors=mean_absolute_error(y_test, predictSalary)
print("mean_absolute_error", errors)

```
